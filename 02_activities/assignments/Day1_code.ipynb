```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0, 100,50)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(x,y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)
```

```python
# Copy from Fabiana
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y, 
marker = '*', 
markersize=12,
color = 'indigo', 
linestyle='--', 
linewidth=2,
markeredgecolor="#cf59fa",  # outline color
markerfacecolor="#FAFAF8")  # fill color
```

```python
# Filled circle with contrasting edge
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y, marker='o', markersize=12, color='indigo', linestyle='--',
        markeredgecolor='black', markerfacecolor='lightblue')
```

```python
# Thick plus sign
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y, marker='D', markersize=7, markeredgewidth=2)
```

```python
# Empty square
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y, marker='o', markersize=7, markerfacecolor='none', 
        markeredgecolor='navy')
```

#### Some examples of markers can be used:
* Plus: ax.plot(x, y, marker='+')<br>
* Cross: ax.plot(x, y, marker='x')<br>
* Thin diamond: ax.plot(x, y, marker='d')<br>
* Pentagon: ax.plot(x, y, marker='p')<br>
* Hexagon: ax.plot(x, y, marker='h')<br>
* Star: ax.plot(x, y, marker='*')<br>
* Filled star: ax.plot(x, y, marker='*', markeredgecolor="none")<br>
* Circle: ax.plot(x, y, marker='o')  <br>
* Square: ax.plot(x, y, marker='s')<br>
* Diamond: ax.plot(x, y, marker='D')<br>
